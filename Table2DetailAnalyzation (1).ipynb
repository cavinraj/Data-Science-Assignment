```python
#Import necessary libraries
import pandas as pd
import numpy as np
```

```python
#read the CSV file
path = "/content/train.csv"
df = pd.read_csv(path)
```

```python
df.head() #display the first rows of our dataframe
```

```python
summary_data = [] #create list to append all our results for table 2 for each statistics eg. percentile,mean,median,mode and etc.
```

```python
if 'Loan_ID' in df.columns:
    df = df.drop('Loan_ID', axis=1)
```

```python
# We want to see frequency for Credit History (1 vs 0) and Term (360 vs 180)
#cols_to_convert = ['Credit_History', 'Loan_Amount_Term']

#for col in cols_to_convert:
    # Convert to string (text) so the code treats it as Categorical
    #df[col] = df[col].astype(str)
    # Note: This will make Mean/Median show "N/A" for these columns,
    # but it will give you the "84% : 16%" frequency you want.
```

```python
summary_data = [] #list to sotre our values

#loop through every column in the dataset
for column in df.columns:

    #basic Info (applies to ALL columns)

    #completeness (How much data is missing?)
    missing_count = df[column].isnull().sum()
    total_rows = len(df)
    completeness_pct = ((total_rows - missing_count) / total_rows) * 100

    #Define which columns we want to FORCE to be Numerical vs Categorical
    #Credit_History (1.0/0.0) -> categorical (Frequency)
    #Loan_Amount_Term (360, 180) -> numerical (Mean/Median)

    is_numerical = False #default to false

    #if it's Credit History, force categorical
    if column == 'Credit_History':
        is_numerical = False

    #if it's Loan_Amount_Term, force numerical
    elif column == 'Loan_Amount_Term':
        is_numerical = True

    #for everything else, let python decide based on the data type
    elif pd.api.types.is_numeric_dtype(df[column]):
        is_numerical = True

    #process based on type

    if is_numerical:
        #calculate math stats
        mean_val = df[column].mean()
        median_val = df[column].median()
        std_dev = df[column].std()
        variance = df[column].var()
        skewness = df[column].skew()
        kurtosis = df[column].kurt()

        #percentiles
        p25 = df[column].quantile(0.25)
        p50 = df[column].quantile(0.50)
        p75 = df[column].quantile(0.75)

        #get mode (Most common number)
        if len(df[column].mode()) > 0:
            mode_val = df[column].mode()[0]
        else:
            mode_val = "N/A"

        #frequency distribution is not for numbers
        freq_dist = "N/A (Numerical)"

    else:
        #calculate frequency (Percentages)
        temp_col = df[column].astype(str) #convert to text temporarily
        counts = temp_col.value_counts(normalize=True) * 100

        dist_str = ""
        #if too many values, take top 3 + "etc"
        if len(counts) > 10:
            loop_data = counts.head(3)
            add_etc = True
        else:
            loop_data = counts
            add_etc = False

        for name, pct in loop_data.items():
            if name.lower() == 'nan': continue
            dist_str = dist_str + f"{name}: {round(pct, 1)}%, "

        if add_etc:
            dist_str += "etc."

        freq_dist = dist_str

        #get mode
        if len(df[column].mode()) > 0:
            mode_val = df[column].mode()[0]
        else:
            mode_val = "N/A"

        #set math stats to N/A
        mean_val = "N/A"
        median_val = "N/A"
        std_dev = "N/A"
        variance = "N/A"
        skewness = "N/A"
        kurtosis = "N/A"
        p25 = "N/A"
        p50 = "N/A"
        p75 = "N/A"

    #formatting helper: simple function to round if it's a number
    def format_num(x):
        if isinstance(x, (int, float)):
            return round(x, 2)
        return x

    #append to list
    summary_data.append({
        "Variable": column,
        "Frequency (Mode)": freq_dist,
        "25th Percentile": format_num(p25),
        "50th Percentile": format_num(p50),
        "75th Percentile": format_num(p75),
        "Completeness (%)": format_num(completeness_pct),
        "Mean": format_num(mean_val),
        "Median": format_num(median_val),
        "Mode": mode_val,
        "Standard Deviation": format_num(std_dev),
        "Variance": format_num(variance),
        "Skewness": format_num(skewness),
        "Kurtosis": format_num(kurtosis)
    })
```

```python
#convert our list into a Pandas DataFrame to display it nicely
stats_df = pd.DataFrame(summary_data)
pd.set_option('display.max_colwidth', None) # Show full text
```
