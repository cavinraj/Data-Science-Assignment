```python
#Import necessary libraries
import pandas as pd
import numpy as np
```

```python
#read the CSV file
path = "/content/post_feature_engineering_2 (1).csv"
df = pd.read_csv(path)
```

```python
summary_data = [] #create list to append all our results for table 2 for each statistics eg. percentile,mean,median,mode and etc.
```

```python
#remove Unnamed_0 its the same as Loan_ID jsut indicating number of rows (Not needed for analysis)
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)
```

```python
# 2. DEFINE COLUMN TYPES
# Explicitly defining which are Categorical
categorical_cols = [
    'dependents',
    'education',
    'credit_history',
    'property_area',
    'loan_status',
    'self_employed'
]

# 3. GENERATE STATISTICS ROWS
table_rows = []

for col in df.columns:
    series = df[col]
    total_len = len(df)
    non_null_count = series.count()

    # Calculate Completeness
    completeness = round((non_null_count / total_len) * 100, 2)
    completeness_str = f"{completeness}%"

    # Determine Data Type
    if col in categorical_cols:
        data_type = "Categorical"
    else:
        data_type = "Numerical"

    # --- CONDITIONAL LOGIC ---
    if data_type == "Numerical":
        # Frequency (Count for Numerical)
        freq_val = non_null_count

        # Stats
        mean_val = round(series.mean(), 2)
        median_val = round(series.median(), 2)
        std_dev = round(series.std(), 2)
        variance = round(series.var(), 2)
        skewness = round(series.skew(), 2)
        kurtosis = round(series.kurt(), 2)

        # Range
        min_v = round(series.min(), 2)
        max_v = round(series.max(), 2)
        range_val = f"{min_v} - {max_v}"

        # Mode
        mode_v = series.mode()
        mode_val = round(mode_v[0], 2) if len(mode_v) > 0 else "N/A"

        # --- NEW: PERCENTILES (25th, 50th, 75th) ---
        p25 = round(series.quantile(0.25), 2)
        p50 = round(series.quantile(0.50), 2)
        p75 = round(series.quantile(0.75), 2)
        percentile_str = f"{p25},{p50},{p75}"

    else:
        # --- CATEGORICAL LOGIC ---
        # 1. Range / Values (Unique sorted)
        unique_vals = sorted(series.dropna().unique())
        range_val = ", ".join([str(x) for x in unique_vals])

        # 2. Frequency (Value : Percentage)
        val_counts_pct = series.value_counts(normalize=True)
        freq_parts = []
        for val in unique_vals:
            pct = round(val_counts_pct[val] * 100, 1)
            freq_parts.append(f"{val} : {pct}%")
        freq_val = ", ".join(freq_parts)

        # Categorical Stats (N/A for numerical metrics)
        mean_val = "-"
        median_val = "-"
        std_dev = "-"
        variance = "-"
        skewness = "-"
        kurtosis = "-"
        percentile_str = "-" # Percentiles don't apply to categorical

        # Mode (Most Frequent Value)
        mode_v = series.mode()
        mode_val = mode_v[0] if len(mode_v) > 0 else "N/A"

    # Append Row
    table_rows.append({
        "Variable": col,
        "Frequency": freq_val,
        "Percentiles (25th,50th,75th)": percentile_str,
        "Data Completeness": completeness_str,
        "Mean": mean_val,
        "Median": median_val,
        "Mode": mode_val,
        "Standard Deviation": std_dev,
        "Variance": variance,
        "Skewness": skewness,
        "Kurtosis": kurtosis
    })
```

```python
#convert our list into a Pandas DataFrame to display it nicely
cleaned_table2_analysation_df = pd.DataFrame(table_rows)
pd.set_option('display.max_colwidth', None) # Show full text
```
